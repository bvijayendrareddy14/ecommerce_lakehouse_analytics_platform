```python
from datetime import datetime

def get_run_id():
    try:
        return str(dbutils.jobs.taskContext().runId())
    except:
        return f"MANUAL_{datetime.now().strftime('%Y%m%d%H%M%S')}"
```

```python
def log_run(run_id, pipeline_name, task_name, status, message=None, error_message=None):

    message = message.replace("'", "") if message else None
    error_message = error_message.replace("'", "") if error_message else None

    spark.sql(f"""
        INSERT INTO commerce_main_dev.util.run_log
        VALUES (
            '{run_id}',
            '{pipeline_name}',
            '{task_name}',
            '{status}',
            {f"'{message}'" if message else "NULL"},
            {f"'{error_message}'" if error_message else "NULL"},
            current_timestamp()
        )
    """)
```
